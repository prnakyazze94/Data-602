```python
class Reverse_String:
    def reverse_words(self, text: str) -> str:
        """Splits the input string into words and joins them in reverse order."""
        words = text.split()
        return " ".join(reversed(words))


# Example usage:
reverser = Reverse_String()
input_str = "hello .py"
output_str = reverser.reverse_words(input_str)

print(f"Input: {input_str}")
print(f"Output: {output_str}")
# Expected Output: .py hello
```

```python
import math


class Circle:
    def __init__(self, radius: float):
        """Initializes the circle with a given radius."""
        self.radius = radius

    def area(self) -> float:
        """Computes and returns the area of the circle (pi * r^2)."""
        return math.pi * (self.radius**2)

    def perimeter(self) -> float:
        """Computes and returns the perimeter/circumference (2 * pi * r)."""
        return 2 * math.pi * self.radius


# Example usage:
circle = Circle(radius=5)

print(f"Radius: {circle.radius}")
print(f"Area: {circle.area():.2f}")
print(f"Perimeter: {circle.perimeter():.2f}")
```

```python
import math


class Circle:

  def __init__(self, radius: float):
    """Initializes the circle with a given radius."""
    self.radius = radius

  def area(self) -> float:
    """Computes and returns the area of the circle (pi * r^2)."""
    return math.pi * (self.radius**2)

  def perimeter(self) -> float:
    """Computes and returns the perimeter/circumference (2 * pi * r)."""
    return 2 * math.pi * self.radius


# Ask for user input:
user_radius = float(input("Enter the radius of the circle: "))
circle = Circle(radius=user_radius)

print(f"Radius: {circle.radius}")
print(f"Area: {circle.area():.2f}")
print(f"Perimeter: {circle.perimeter():.2f}")

```
