```python
# Question 1 Create a list called animals
animals = ["cat", "dog", "manta ray", "horse", "crouching tiger"]

# Question 2  loop through and print each item in the animal list by iterating through an index number and using range().
#Set a variable len_animals to the length of the animal list.
len_animals = len(animals)

for i in range(len_animals):
    print(animals[i])
```

```python
countdown = [9, 8, 7, 5, 4, 2, 1, 6, 10, 3, 0, -5]

# Sort the list in descending order
countdown.sort(reverse=True)

# The 5th element corresponds to index 4 (because indexing starts at zero)
the_fifth_element = countdown[4]

print(f"Sorted countdown: {countdown}")
print(f"The 5th element: {the_fifth_element}")
```

```python
list1 = [10, 20, [300, 400, [5000, 6000], 500], 30, 40]

# Access the inner nested list [5000, 6000] and append 7000
list1[2][2].append(7000)

print(list1)
```

```python
# 5. Write a program to remove all occurrences of item 20 in the following list.
list2 = [5, 20, 30, 15, 20, 30, 20]

# Remove every instance of 20 using a while loop
while 20 in list2:
    list2.remove(20)

print(list2)
```
