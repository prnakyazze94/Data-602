**Question 3: Nested Logic for Asset Depreciation**

Your firm has a new tax rule for software and hardware assets.
* If the asset is "Hardware":
  * and it costs 5,000 or more, it gets a $1000 deprecation bonus.
  
  * And it costs less than 5,000, it gets a $250 depreciation bonus.

* If the asset is "Software":
  * It always gets a flat $500 depreciation bonus regardless of cost.
* For any other asset type, print "Invalid asset class."

Write a program that takes the **asset type** and **cost** as inputs, uses nested `if`/`elif` statements to determine the bonus, and prints the result.

```python
# Write your Depreciation Bonus Calculator here
asset_type = str(input("enter assest type: "))
cost = float(input("enter cost: "))

if asset_type == "Hardware" :
  if cost >= 5000 :
    print("$1000 deprecation bonus.")
  else :
    print("$250 deprecation bonus.")
elif asset_type == "Software" :
  print("$500 deprecation bonus.")
else:
  print("invalid asset class")
```

**Question 4: The Payroll Audit Challenge (Group Challenge)**

Your group needs to write a script to calculate weekly payroll for hourly data entry clerks, factoring in overtime and a standard tax withholding.

**Requirements:**
* Ask for `hours_worked` and `hourly_rate`.
* Regular pay applies up to 40 hours.
* Overtime pay (1.5x the hourly rate) applies to any hours worked *over* 40.
* After calculating the Gross Pay (Regular + Overtime), deduct a flat 20% for taxes.
* Print a clean "Pay Stub" showing: Gross Pay, Taxes Withheld, and Net Pay (formatted to two decimal places).

```python
# Ask for inputs
hours_worked = float(input("enter hours worked: "))
hourly_rate = float(input("enter hourly_rate: "))

# Calculate Gross Pay (Handle overtime logic)
if hours_worked <= 40 :
  grosspay = hours_worked*hourly_rate
else:
  ot_hours = hours_worked - 40

  wage1 = hourly_rate*40
  wage2 = ot_hours* (hourly_rate*1.5)
  grosspay = wage1 + wage2

# Calculate Taxes and Net Pay
taxes_withheld = grosspay * .2
net_pay = grosspay - taxes_withheld
# Print the Pay Stub
print(f"Paystub:\n\tGross Pay:       ${grosspay:.2f}\n\tTaxes Withheld:  ${taxes_withheld:.2f}\n\tNet Pay:         ${net_pay:.2f}")
```
