```python
2. Design a program that asks the user to enter a store’s sales for each day of the
week. The amounts should be stored in a list. Use a loop to calculate the total sales for
the week and display the result.
```

```python
def sales():
    days = [
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday",
    ]
    daily_sales = []#The amounts stored in a list.

    # Collect sales for each day of the week
    for day in days:
        amount = float(input(f"Enter sales for {day}: $"))
        daily_sales.append(amount)

    # Use a loop to calculate the total sales
    total_sales = 0.0
    for sale in daily_sales:
        total_sales += sale

    # Display the total sales formatted as currency
    print(f"\nTotal sales for the week: ${total_sales:,.2f}")


if __name__ == "__main__":
    sales()
```

 Write a program that creates a dictionary containing course numbers and the room
numbers of the rooms where the courses meet.

```python
# Create a dictionary containing course numbers to room numbers
course_rooms = {
    "Chem101": "Room 101",
    "Eng111": "Room 204",
    "Math104": "Room 315",
    "Business103": "Room 402",
    "Art100": "Studio B",
}
#  course numbers and the names of the instructors that teach each course
course_instructors = {
    "Chem101": "Dr. Smith",
    "Eng111": "Prof. Davis",
    "Math104": "Dr. Taylor",
    "Business103": "Prof. Anderson",
    "Art100": "Ms. Martinez",
}
# course numbers and meeting times
course_times = {
    "Chem101": "8:00 a.m.",
    "Eng111": "9:00 a.m.",
    "Math104": "10:00 a.m.",
    "Business103": "11:00 a.m.",
    "Art100": "1:00 p.m.",
}

# Prompt the user for a course number
user_input = input("Enter a course number (e.g., Chem101, Eng111, Math104): ").strip()

# Find the matching dictionary key regardless of case
matched_key = None
for key in course_rooms:
    if key.casefold() == user_input.casefold():
        matched_key = key
        break

# Look up and display details
if matched_key:
    print(f"\nCourse Details for {matched_key}:")
    print(f"  Room Number:  {course_rooms[matched_key]}")
    print(f"  Instructor:   {course_instructors[matched_key]}")
    print(f"  Meeting Time: {course_times[matched_key]}")
else:
    print(f"\nCourse '{user_input}' was not found.")
```

 Write a program that keeps names and email addresses in a dictionary as
key-value pairs.

```python
contacts = {
    "Alice Johnson": "alice@gmail.com",
    "Bob Smith": "bob@yahoo.com",
    "Charlie Brown": "charlie@hotmail.com",
}
def display_menu():
    print("\n--- Names and Email Address Book ---")
    print("1. Look up an email")
    print("2. Add a new contact")
    print("3. Change an existing email")
    print("4. Delete a contact")
    print("5. Quit")


while True:
    display_menu()
    choice = input("\nEnter your choice (1-5): ").strip()

    if choice == "1":
        name = input("Enter the name to look up: ").strip()
        print(f"Email: {contacts.get(name, 'Contact not found.')}")

    elif choice == "2":
        name = input("Enter contact name: ").strip()
        if name in contacts:
            print("That contact already exists.")
        else:
            email = input("Enter email address: ").strip()
            contacts[name] = email
            print(f"{name} added successfully.")

    elif choice == "3":
        name = input("Enter contact name: ").strip()
        if name in contacts:
            new_email = input("Enter the new email address: ").strip()
            contacts[name] = new_email
            print(f"Email updated for {name}.")
        else:
            print("Contact not found.")

    elif choice == "4":
        name = input("Enter the name to delete: ").strip()
        if name in contacts:
            del contacts[name]
            print(f"{name} deleted.")
        else:
            print("Contact not found.")

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid option. Please enter a number between 1 and 5.")
```
