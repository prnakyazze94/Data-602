Write a function to calculate the area and perimeter of a rectangle.

```python
Write a function to check if a number is even or not.  The function should indicate to the user even or odd
```

```python
def rectangle(length: float, width: float) -> tuple[float, float]:
    """Calculate the area and perimeter of a rectangle.


    """
    if length < 0 or width < 0:
        raise ValueError("Dimensions must be non-negative.")

    area = length * width
    perimeter = 2 * (length + width)

    return area, perimeter


# Example usage:
length = float(input("Enter Lenght:"))
width = float(input("Enter Lenght:"))

area, perimeter = rectangle(length, width)
print(f"Area of a rectangle is: {area}")
print(f"Perimeter of a rectangle is: {perimeter}")
```

Write a function to check if a number is even or not.  The function should indicate to the user even or odd

```python
def odd_even(number: int) -> str:
    """if you get no remainder when you divide a number the it's eveb.


    """
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"


# Example usage:
user_input = int(input("Enter any numberr: "))
result = odd_even(user_input)
print(f"The number {user_input} is {result}.")
```

Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```python
def count_case_in_word(text: str) -> tuple[int, int]:
    """Calculate the number of uppercase and lowercase letters in a string.


    """
    upper_count = 0
    lower_count = 0

    for char in text:
        if char.isupper():
            upper_count += 1
        elif char.islower():
            lower_count += 1

    return upper_count, lower_count


# Example usage:
user_text = input("Enter a word: ")
upper, lower = count_case_in_word(user_text)

print(f"Original String: {user_text}")
print(f"No. of Upper case characters: {upper}")
print(f"No. of Lower case characters: {lower}")
```

Write a Python function to sum all the numbers in a list

```python
def sum_list(numbers: list[int | float]) -> int | float:
    """Return the sum of all numbers in a list."""
    return sum(numbers)


# Example usage:
nums = [8, 2, 3, 0, 7]
print(sum_list(nums))  # Output: 20
```

 Write a Python program to create a function that takes one argument, and that argument will be multiplied with an unknown given number.

```python
import random

def make_multiplier(n: float):
    return lambda x: x * n

# 1. Pick and inspect the random number
chosen_factor = random.randint(1, 10)
random_mult_1 = make_multiplier(chosen_factor)

# 2. Test
user_val = 10
result = random_mult_1(user_val)

print(f"Random factor was: {chosen_factor}")
print(f"{user_val} * {chosen_factor} = {result}")
```
