 What is data aggregation?  Give an example in Python.


 Data aggregation is the process of taking multiple values and summarizing them into a single representative value (such as a sum, mean, minimum, maximum, or count).

```python


df4 = pd.DataFrame({'scores': [88, 92, 79, 95, 85]})

# Aggregating a column to a single metric
mean_score = df4['scores'].mean()
total_score = df4['scores'].sum()

# Aggregating multiple metrics at once
summary = df4['scores'].agg(['mean', 'sum', 'min', 'max'])
print(summary)
```

groupby() in Pandas implements the Split Apply Combine pattern:

Split: Divides a DataFrame into distinct subsets based on unique values in one or more specified categorical columns.

Apply: Computes an operation or statistical summary independently on each subgroup (e.g., .mean(), .sum(), .count(), .min(), .max(), or custom transformations via .agg()).

Combine: Merges all intermediate subgroup results back together into a single unified Series or DataFrame.

```python


# 1. Ensure the score columns are properly typed as numeric floats
df3['Education Score'] = pd.to_numeric(df3['Education Score'], errors='coerce')
df3['Overall Score'] = pd.to_numeric(df3['Overall Score'], errors='coerce')

# 2. Filter for states in the 'West' region
west_df = df3[df3['Region'] == 'West']

# 3. Preview West region records
print("=== West Region States ===")
print(west_df[['Rank', 'State', 'Education Score', 'Overall Score', 'Region']])

# 4. GroupBy West region to verify average numeric scores
west_mean = (
    west_df.groupby('Region')[['Education Score', 'Overall Score']]
    .mean()
    .round(2)
)
print("\n=== Mean Scores for West Region ===")
print(west_mean)
```

The mean education score in the West is 43.79
