Series: A 1D labeled array capable of holding any data type.

DataFrame: A 2D labeled, tabular data structure with columns of potentially different types (similar to an SQL table or Excel spreadsheet).

Write a Pandas program to compare the elements of the two Pandas Series.
Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 10]

```python


s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 10])

print("Equal:")
print(s1 == s2)

print("\nGreater than:")
print(s1 > s2)

print("\nLess than:")
print(s1 < s2)
```
